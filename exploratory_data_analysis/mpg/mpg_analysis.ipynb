```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("mpg.csv")
df.head()
```

```python
df.info()
```

```python
# '?' karakterini NaN ile değiştir, sonra float'a çevir
df["horsepower"] = df["horsepower"].replace('?', np.nan).astype(float)
```

```python
df.shape
```

```python
df.describe().T
```

```python
df.isna().sum()
```

```python
df.columns
```

```python
df.model_year.unique()
```

```python
df.model_year.value_counts().plot(kind = "bar", color = "steelblue")
plt.xlabel("Model Year")
plt.ylabel("Count")
plt.title("Distribution of Model Years in MPG Dataset")
plt.show()
```

```python
df.groupby("model_year")["mpg"].mean().plot(kind = "line", marker = "o", color = "orange")
```

```python
df.groupby("model_year")["cylinders"].mean().plot(kind = "line", marker = "o", color = "orange")
```

```python
df.columns
```

```python
df.groupby("model_year")["acceleration"].mean().plot(kind = "line", marker = "o", color = "orange")
```

```python
plt.figure(figsize = (9, 5))
sns.heatmap(df.corr(numeric_only=True), annot=True)
```
