```python
df_wq.corr()
```

```python
plt.figure(figsize = (12, 6))
sns.heatmap(df_wq.corr(), annot = True)
```

```python
# density ile fixed acidity arasinda pozitif yonlu guclu bir iliski var gibi gorunuyor.
# citric acis ile fixed acidity arasinda da pozitif yonlu guclu bir iliski var.
# total sulfur dioxide ile free sulfur dioxide arasinda da pozitif yonlu guclu bir iliski var.
# ph ile fixed acidity arasinda negatif yonlu orta guclukta bir iliski var.
```

```python
sns.set_style("whitegrid")
sns.displot(x = "pH", data = df_wq, color = "purple", kde = True) # ph degerlerinin cogunlukla 3.2-3.4 araliginda oldugunu goruyoruz
```

```python
sns.histplot(x = "alcohol", data = df_wq, color = "purple") # alkol oranlarinin cogunlukla 10-12 araliginda oldugunu goruyoruz
```

```python
sns.pairplot(df_wq) # iliskileri daha iyi gozlemleyebilmek icin pairplot kullanabiliriz
```

```python
sns.boxplot(x = "quality", y = "alcohol", data = df_wq) # kalite arttikca alkol oraninin da arttigini goruyoruz
```

```python
sns.scatterplot(x ="density", y = "fixed acidity", data = df_wq, hue = "quality") 
# density ile fixed acidity arasinda pozitif yonlu guclu bir iliski var gibi gorunuyor
# density degeri azaldikca quality degeri artiyor gibi gorunuyor
```
